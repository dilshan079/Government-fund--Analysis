```python
df["Utilization_Percent"]=df["Utilization_Percent"].fillna(df["Utilization_Percent"].mean().round())
```

```python
df["Utilization_Percent"].mean()
```

```python
df["Fund_Utilized_Lakhs"] = df["Fund_Utilized_Lakhs"].fillna(
    (df["Fund_Released_Lakhs"] * df["Utilization_Percent"] / 100).round()
)
```

```python
df["Physical_Achievement_Units"] = df["Physical_Achievement_Units"].fillna(
    (df["Physical_Target_Units"] * df["Utilization_Percent"] / 100).round()
)
```

```python
df["Beneficiaries_Covered"] = df["Beneficiaries_Covered"].fillna(
    (df["Beneficiaries_Target"] * df["Utilization_Percent"] / 100).round()
)
```

```python
df["Audit_Remarks"]=df["Audit_Remarks"].fillna(df["Audit_Remarks"].mode()[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x="Status")

plt.title("Distribution of Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter only Delayed status
delayed_df = df[df["Status"] == "Delayed"]

# Bar plot of Audit Remarks
plt.figure(figsize=(10, 6))

sns.countplot(
    data=delayed_df,
    x="Audit_Remarks"
)

plt.title("Audit Remarks for Delayed Projects")
plt.xlabel("Audit Remarks")
plt.ylabel("Number of Projects")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python

released = df["Fund_Released_Lakhs"].sum()
utilized = df["Fund_Utilized_Lakhs"].sum()

plt.figure(figsize=(8, 5))

bars = plt.bar(
    ["Fund Released", "Fund Utilized"],
    [released, utilized]
)

plt.ylabel("Amount (₹ Lakhs)")
plt.title("Total Fund Released vs Total Fund Utilized")

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{bar.get_height():,.2f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python


yearly_fund = df.groupby("Financial_Year")["Fund_Released_Lakhs"].sum()

plt.figure(figsize=(10, 5))

bars = plt.bar(yearly_fund.index, yearly_fund.values)

plt.xlabel("Financial Year")
plt.ylabel("Fund Released (₹ Lakhs)")
plt.title("Fund Released by Financial Year")
plt.xticks(rotation=45)

# Display values
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{bar.get_height():,.2f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

yearly_funds = df.groupby("Financial_Year")[
    [
        "Sanctioned_Amount_Lakhs",
        "Fund_Released_Lakhs",
        "Fund_Utilized_Lakhs"
    ]
].sum()

ax = yearly_funds.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title("Sanctioned, Released and Utilized Funds by Financial Year")
plt.xlabel("Financial Year")
plt.ylabel("Amount (₹ Lakhs)")
plt.xticks(rotation=45)
plt.legend(
    title="Fund Type",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```
